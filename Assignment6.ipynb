```python
import pandas as pd
data= {
    "Name":["Ali","shaan","King","Ali"],
    "Marks":[85,60,90,85],
    "City":["Pune","Mumbai","Delhi","Pune"]}
df=pd.DataFrame(data)
#Display original dataset
print("Original dataset:")
print(df)
#Display first five rows
print("\nFirst Five Rows!")
print(df.head())
```

```python
#Display dataset information
print("\nDataset Information:")
df.info()
```

```python
#Display statiscal summary
print("\n Statiscal summary:")
print(df.describe())
```

```python
#Check missing value
print("\n missing values:")
print(df.isnull().sum())
```

```python
#Fill the missing marks with the mean
df["Marks"]
df["Marks"].fillna(df["Marks"].mean())
print(df)
```

```python
#Identify duplicate records
print("\n Duplicate Records:")
print(df.duplicated())
```

```python
#Remove the duplicated valuye
df=df.drop_duplicates()
```

```python
#Select students who scored more than 80
print("\n Students scoring above 80:")
print(df.loc[df["Marks"]>80])
```

```python
#Display cleaned dataset
print("\nCleaned Dataset:")
print(df)
```
