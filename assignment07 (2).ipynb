```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV dataset
df = pd.read_csv(r'C:\Users\Admin\Desktop\sample_data_assignment_07.csv')

# Display the dataset
print(df.head())
```

```python
#Display the dataset
print("Dataset")
print(df)
```

```python
#line plot
plt.figure(figsize=(7,4.5))
plt.plot(df["Student"],df["Python"],marker="o")
plt.title("Python marks of students")
plt.xlabel("Student")
plt.ylabel("Python marks")
plt.grid(True)
plt.show()
```

```python
#bar plot
plt.figure(figsize=(7,4.5))
plt.bar(df["Student"],df["Maths"])
plt.title("Maths marks of students")
plt.xlabel("Student")
plt.ylabel("Python marks")
plt.grid(True)
plt.show()
```

```python
#Scatter plot
plt.figure(figsize=(7,4.5))
plt.scatter(df["Maths"],df["Python"],s=70)
plt.title("Maths rs of students")
plt.xlabel("Maths Marks")
plt.ylabel("Python marks")
plt.grid(True)
plt.show()
```

```python
#correlation MATRIX AND HEATMAP
correlation=df[["Maths","Python","Attendance"]].corr()
print("Correlation Matrix")
print(correlation)
plt.figure(figsize=(6.5,6))
sns.heatmap(correlation,annot=True,cmap="Reds",vmin=-1,vmax=1)
plt.title("correlation Heatmap")
plt.show()
```
